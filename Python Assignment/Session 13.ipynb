```python
def reverse_string(s):
    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]

# Test
text = "hello"
print("Original String:", text)
print("Reversed String:", reverse_string(text))
```

```python
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])

# Playlist durations (seconds)
playlist = [210, 180, 240, 150]

print("Total Duration:", sum_playlist_durations(playlist), "seconds")
```

```python
count = 10
def update_count():
    count = 5
    print('Inside:', count)

update_count()
print('Outside:', count)

"""Explanation
count = 10 → Global variable
count = 5 inside update_count() → Local variable
The local variable exists only inside the function.
It does not change the global variable."""
```

```python
def count_likes(post):
    total = post["likes"]

    for reply in post.get("replies", []):
        total += count_likes(reply)

    return total

# Sample nested dictionary
posts = {
    "likes": 100,
    "replies": [
        {
            "likes": 50,
            "replies": [
                {"likes": 20, "replies": []},
                {"likes": 30, "replies": []}
            ]
        },
        {
            "likes": 40,
            "replies": []
        }
    ]
}

print("Total Likes:", count_likes(posts))
```

```python
app_status = "Online"

def check_status():
    # Local variable
    user_status = "Typing..."

    print("Inside Function")
    print("App Status:", app_status)
    print("User Status:", user_status)

print("Before Function")
print("App Status:", app_status)

check_status()

print("After Function")
print("App Status:", app_status)
```
