```python
def clean_brand_name(name):
    name = name.strip().replace("-"," ")
    return name    
clean_brand_name("oneplus-Nord")
```

```python
def Brand_namde(name):
    name = name.split(" ",1)
    return name

Brand_namde('Apple iPhone 14 Pro Max')
```

```python
def product_name(name, price):
    print(f"{name}-{price}")

product_name('Boat Earbuds', "1299")
```

```python
Brands = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']
clean_list = []
for a in Brands:
    c = a.strip().replace("-"," ").title()
    clean_list.append(c)

print(clean_list)
```
