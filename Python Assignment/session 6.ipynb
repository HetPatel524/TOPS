```python
playlist_prices={"hindi song":2000,"punjabi _song":2100,"gujrati song":1000,"english sing":900,"japanese song":3000}

def update_playlist_price(playlist, new_price):
    playlist_prices[playlist]=new_price

playlist_name=str(input("Enter a play list name:"))
playlist_new_prices=int(input("Enter a list new price:"))
update_playlist_price(playlist_name,playlist_new_prices)

print(playlist_prices)
```

```python
def delt_playlist(playlist_names):
    del playlist_prices[playlist_names]
    print(playlist_prices)

playlist_name=str(input("Enter a play list name:"))
delt_playlist(playlist_name)
```
