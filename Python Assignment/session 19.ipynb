```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration

    def play_preview(self):
        print(f"Playing 30-second preview of {self.title} by {self.artist}")

# Create object
song1 = Song("Blinding Lights", "The Weeknd", 200)

# Call the method
song1.play_preview()
```

```python
class FoodOrder:
    def __init__(self, restaurant_name):
        self.restaurant_name = restaurant_name
        self.items = []
        self.total_price = 0

    def add_item(self, item, price):
        self.items.append(item)
        self.total_price += price

# Create object
order = FoodOrder("Pizza Hut")

# Add two items
order.add_item("Veg Pizza", 299)
order.add_item("Garlic Bread", 149)

# Display order details
print("Restaurant:", order.restaurant_name)
print("Items:", order.items)
print("Total Price: ₹", order.total_price)
```

```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration
        self.play_count = 0

    def increment_play_count(self):
        self.play_count += 1

# Create object
song1 = Song("Blinding Lights", "The Weeknd", 200)

# Increase play count
song1.increment_play_count()
song1.increment_play_count()
song1.increment_play_count()

# Display play count
print("Song:", song1.title)
print("Play Count:", song1.play_count)
```
