```python
def calculate_final_price(price, discount_rate):
    final_pric = price - price*discount_rate/100
    return final_pric

price = int (input("Enter a price"))
discount_rate = float(input("Enter a discount rate:"))

calculate_final_price(price,discount_rate)
```

```python
def apply_discount(price, rate=10):
    print(f"your price is {price} and your discount rate {rate}% and Final amount is : {price-price*rate/100}")

apply_discount(2000,20)
apply_discount(2000)
```

```python
def calculate_cashback(amount, cashback_rate=5):
    print(f"your price is {amount} and cashback rate is {cashback_rate}% your cashback amount is {amount*cashback_rate/100} ")
    
print("zometo",calculate_cashback(2000))
print("flipkart",calculate_cashback(2000,7))
```
